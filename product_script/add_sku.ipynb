```python
#!/usr/bin/env python3
import csv
import sys
from pathlib import Path


def normalize_header(value: str) -> str:
    return (value or "").strip().lower().replace("_", " ").replace("/", " ").replace("-", " ")


def find_product_service_name_column(fieldnames):
    if not fieldnames:
        return None

    normalized_map = {normalize_header(name): name for name in fieldnames}

    aliases = [
        "product service name",
        "product/service name",
        "product name",
    ]
    for alias in aliases:
        if alias in normalized_map:
            return normalized_map[alias]

    for name in fieldnames:
        n = normalize_header(name)
        if "product" in n and "service" in n and "name" in n:
            return name
        if "product" in n and "name" in n and "service" not in n:
            return name

    return None


def extract_products(csv_path: str):
    values = []

    with open(csv_path, "r", newline="", encoding="utf-8-sig") as f:
        reader = csv.DictReader(f)
        if reader.fieldnames is None:
            raise ValueError(f"CSV file '{csv_path}' has no header row.")

        for row in reader:
            value = (row.get("Product/Service Name") or "").strip()
            sku = (row.get("SKU") or "").strip()
            values.append([value, sku])

    return values


```

```python
product_name = extract_products("./inventory.csv")
```
